```python
def us(*friends):
  for i in friends:
    print(f"Hey {i}")

us("Tanvi" ,"Madhu" ,"Neha")
```

```python
def us(**friends):
  for key,value in friends.items():
    print(f"{key} : {value}")

us(Madhu = 92 , Tanvi = 86)
```

```python
from functools import reduce
square = lambda x : x * x
numbers = [1,2,3]
result = list(map(square,numbers))
output = list(filter(lambda x : x%2==0,result))
final = reduce(lambda x, y: x + y, result)

print(output)
print(final)
```
