```python
import pandas as pd
student_data = {
    "Student" : ['S1','S2','S3','S4','S5','S6','S7','S8','S9','S10'],
    "Study_Hours" : [2,3,4,5,6,7,8,4,6,9],
    'Attendence' : [65,70,75,80,85,90,92,78,88,95],
    'Assignment_Score' : [55,60,65,72,78,85,88,70,80,92],
    'Previous_Score' : [58,62,68,70,76,82,86,72,79,90],
    'Final_Score' : [60,64,70,75,82,88,91,74,84,94]
}

df = pd.DataFrame(student_data)
df
print(df[["Study_Hours",'Attendence','Assignment_Score','Previous_Score','Final_Score']].mean())
print(df['Final_Score'].max())
print(df['Final_Score'].min())
print(df[df["Attendence"] > 80]["Student"])
print(df.describe())
```

```python
import numpy as np

def L(x):
    return 5*x - 0.2*x**2

x = np.linspace(0, 5, 1000)
y = L(x)

area = np.trapezoid(y, x)

print("Numerical approximation:", area)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

correlation = df.corr(numeric_only=True)

print("Correlation Matrix:")
print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
print("Correlation with Final Score:")
print(correlation["Final_Score"])
```

```python
correlation_with_final = correlation["Final_Score"].drop("Final_Score")

strongest_variable = correlation_with_final.abs().idxmax()
strongest_value = correlation_with_final[strongest_variable]

print("Strongest relationship:", strongest_variable)
print("Correlation:", strongest_value)
```

```python
import numpy as np

X = df[["Study_Hours", "Attendence", "Assignment_Score"]].values
Y = df["Final_Score"].values

# Add a column of 1s for the intercept
X = np.column_stack((np.ones(len(X)), X))

print("X matrix:")
print(X)

print("Y:")
print(Y)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
```
