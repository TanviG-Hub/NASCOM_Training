```python
data = {
    "0" : [32,400,20,95],
    "1" : [35,395,22,88],
    "2" : [38,390,25,80],
    "3" : [40,385,28,70],
    "4" : [42,380,30,60],
    "5" : [45,375,32,48],

}

for time , values in data.items():
    print("Time:" , time)
    print("Temperature: ",values[0])
    print("Voltage: ", values[1])
    print("Current: ",values[2])
    print("Charge: ",values[3])
    print("\n")
```

```python
for time,values in data.items():
  if(values[0] > 40):
      print("High Temperature Alert")
  else:
    print("Temperature Normal")
```

```python
for time,values in data.items():
  if(values[3] < 50):
      print("“Low Battery Alert”")
  else:
    print("“Battery Charge Normal”")
```

```python
def battery_status(charge):
        if(charge > 70):
            print("High")
        elif(50 <= charge <=70):
           print("Medium")
        else:
           print("Low")

for key,values in data.items():
    battery_status(values[3])
    print("\n")
```
